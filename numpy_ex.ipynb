## N09 · 기존 배열을 기준으로 공간 확보: np.empty_like()

PPT 16페이지 · 인쇄 쪽번호 15

```python
# N09 · PPT 16페이지 (인쇄 쪽번호 15) · 기본 예제
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.shape)
result = np.empty_like(arr)
print('초기화 전:', result)
result[:] = arr * 2
print('모든 칸을 계산 결과로 채운 뒤:', result)
```

**바꿔 보면? 정수 배열을 기준으로 실수 결과 공간을 만들려면?**

```python
# N09 · PPT 16페이지 (인쇄 쪽번호 15) · 변형 예제
a=np.array([1,2,3])
b=np.empty_like(a,dtype=float)
print(b)
b[:]= a/2
print(b,b.dtype)
```

dtype을 지정하면 원본의 정수 자료형 대신 실수 자료형을 사용합니다.

## N10 · Identity matrix(단위행렬)

PPT 17페이지 · 인쇄 쪽번호 16

```python
# N10 · PPT 17페이지 (인쇄 쪽번호 16) · 기본 예제
I = np.identity(3)
print('3×3 단위행렬:', I)
A = np.array([2, 3, 4])
print('A @ I:', A @ I)
np.testing.assert_array_equal(A @ I, A)
# 보충: @는 행렬 곱, *는 원소별 곱입니다.
A = np.array([[2, 3, 4], [5, 6, 7], [8, 9, 10]])
print('행렬 곱 A @ I:', A @ I)
print('원소별 곱 A * I:', A * I)
```

**바꿔 보면? 단위행렬 크기를 바꾸면? *로 곱하면?**

```python
# N10 · PPT 17페이지 (인쇄 쪽번호 16) · 변형 예제
A=np.array([[2,3],[4,5]])
I=np.identity(2)
print("@:", A@I)
print("*:", A*I)
```

@는 원래 행렬을 보존합니다. *는 같은 위치끼리 곱하므로 대각선 밖 원소가 0이 됩니다.

```python
a = np.array([True, False], dtype=bool)
print(a.dtype.kind)  # b: 불리언 계열이라는 뜻

b = np.array([1, 2], dtype='b')
print(b.dtype)       # int8: 여기서 'b'는 8비트 정수를 지정
```

## N18 · 자료형 변환에는 조건이 필요

PPT 26페이지 · 인쇄 쪽번호 25

```python
# N18 · PPT 26페이지 (인쇄 쪽번호 25) · 기본 예제
# 의도된 오류: 예외를 잡아 뒤의 셀도 계속 실행되게 합니다.
try:
    arr = np.array(['a', '2', '3'], dtype='i')
    print(arr)
    print(arr.dtype)
except ValueError as error:
    print('예상한 ValueError:', error)
print('변환 가능한 문자열:', np.array(['1', '2', '3'], dtype='i'))
```

**바꿔 보면? 문자열 3.5를 곧바로 정수로 바꿀 수 있을까?**

3.5는 정수 형식 문자열이 아니므로 int 변환은 실패합니다. float 변환은 가능합니다.

## N53 · 자주 사용하는 수학·통계 함수

```python
# N53 · PPT 65페이지 (인쇄 쪽번호 64) · 기본 예제
# 슬라이드의 모든 함수를 실행 가능한 작은 예제로 확인합니다.
arr1 = np.array([7, 10])
arr2 = np.array([3, 4])
print('덧셈:', np.add(arr1, arr2))
print('뺄셈:', np.subtract(arr1, arr2))
print('원소별 곱셈:', np.multiply(arr1, arr2))
print('나눗셈:', np.divide(arr1, arr2))
print('거듭제곱:', np.power(arr1, arr2))
print('부호 반전:', np.negative(arr1))
print('나머지:', np.mod(arr1, arr2))
print('같은 나머지 연산:', np.remainder(arr1, arr2))
quotient, remainder = np.divmod(arr1, arr2)
print('몫:', quotient, '나머지:', remainder)
print('절댓값:', np.absolute([-3, 0, 2]))
arr = np.array([1, 2, 100])
print('산술평균:', np.mean(arr))
print('중앙값:', np.median(arr))
print('표준편차(ddof=0):', np.std(arr))
print('1차원 내적:', np.dot([1, 2], [3, 4]))
A = np.array([[1, 2], [3, 4]])
print('2차원 행렬 곱:', np.dot(A, np.eye(2)))
```

## N56 · Logarithm(로그): 거듭제곱의 지수 구하기

PPT 68페이지 · 인쇄 쪽번호 67

```python
# N56 · PPT 68페이지 (인쇄 쪽번호 67) · 기본 예제
print('밑이 2인 로그:', np.log2(8))
print('밑이 10인 로그:', np.log10(1000))
print('자연로그:', np.log(np.e))
print('밑이 3인 81의 로그:', np.log(81) / np.log(3))
# frompyfunc는 사용자 함수를 ufunc로 감싸는 별도 도구입니다.
# 아래는 작동 확인용 예이며, 내장 NumPy 연산보다 빨라지는 도구가 아닙니다.
def add_one(x):
    return x + 1
add_one_ufunc = np.frompyfunc(add_one, 1, 1)
result = add_one_ufunc(np.array([1, 2, 3]))
print('사용자 함수 적용:', result, 'dtype:', result.dtype)
```

함수가 각 원소에 적용됩니다. 0과 1 사이의 양수는 밑이 1보다 큰 로그에서 음수가 됩니다.

```python
# E10 · PPT 76페이지 (인쇄 쪽번호 75) · 직접 풀기
n = 8
answer = None
# 여기에 풀이를 작성하세요.
# index가 짝수면 0, 홀수면 1인 n*n 크기의 list 생성
l = [1 if i & 1 else 0 for i in range(n * n)]
# array로 먼저 변환 시키고 n*n shape으로 변환
answer = np.array(l).reshape((n,n))
print("n = 8 :")
print(answer)
print("\n"*2)
#n이 5일 때
n = 5
l = [1 if i & 1 else 0 for i in range(n * n)]
answer = np.array(l).reshape((n,n))
print("n = 5 :")
print(answer)
```
