## Q02. 중복 행을 지우고 빈칸 채우기

`raw`는 그대로 두고 작업용 표 `clean`을 만듭니다. 아래 코드를 순서대로 완성하세요.

참고: N18·N20·N21

### Q02-1 · 복사본에서 중복 제거

아래 코드는 `raw`를 복사합니다. 이어서 **완전히 같은 행은 첫 번째만 남기고 나머지를 제거**한 결과를 `clean`에 저장하세요. 마지막에 `clean`의 행 수를 출력하세요.

프로그램이나 학년만 같다는 이유로 다른 학생의 행을 지우면 안 됩니다.

사용할 수 있는 기능: `drop_duplicates()`, `len()`

```python
clean = raw.copy()
# 이어서 중복을 제거하고 결과를 clean에 저장하세요.
clean = clean.drop_duplicates()
# clean의 행 수를 출력하세요.
print("Num of clean row : ",len(clean))
```

### Q02-2 · 이동 시간의 중앙값 구하기

중복을 제거한 `clean`에서 `TravelMinutes`의 중앙값을 구해 **`median_time`**에 저장하고 출력하세요.

빈칸은 중앙값 계산에서 자동으로 제외됩니다.

사용할 수 있는 기능: `median()`

```python
# 여기에 코드를 작성하세요.
median_time = clean["TravelMinutes"].median()
print("Median of TravelMinutes : ",median_time)
```

### Q02-3 · 이동 시간 빈칸 채우기

`clean`의 `TravelMinutes` 빈칸을 앞에서 구한 `median_time`으로 채우세요. 결과는 `clean`의 해당 열에 저장하고, 남은 빈칸 개수를 출력하세요.

이것은 과제에서 정한 대체 규칙이지, 실제 이동 시간을 알아내는 과정은 아닙니다.

사용할 수 있는 기능: `fillna()`, `isna()`, `sum()`

```python
# 여기에 코드를 작성하세요.
print("Before : ",clean["TravelMinutes"].isna().sum())
clean["TravelMinutes"] = clean["TravelMinutes"].fillna(median_time)
print("After : ", clean["TravelMinutes"].isna().sum())
```

### Q02-4 · 알게 된 경로의 빈칸 채우기

`clean`의 `Channel` 빈칸을 문자열 `Unknown`으로 채우세요. 결과는 `clean`의 해당 열에 저장하고, 남은 빈칸 개수를 출력하세요. `Unknown`은 ‘알 수 없음’이라는 뜻입니다.

사용할 수 있는 기능: `fillna()`

```python
# 여기에 코드를 작성하세요.
print("Before : ",clean["Channel"].isna().sum())
clean["Channel"] = clean["Channel"].fillna("Unknown")
print("After : ",clean["Channel"].isna().sum())
```

### Q02-5 · 원자료와 처리 결과 비교

**`clean`의 행 수**, **`clean`의 열별 빈칸 개수**, **`raw`의 열별 빈칸 개수**를 각각 출력하세요.

```python
# 여기에 코드를 작성하세요.
print("Num of clean row : ",len(clean))
print("\n\nNum of None in each row of clean")
for attribute in clean.columns:
    print(attribute," : ",clean[attribute].isna().sum())
print("\n\nNum of None in each row of raw")
for attribute in raw.columns:
    print(attribute," : ",raw[attribute].isna().sum())
```
